```python
from unified_planning.io import PDDLReader
from unified_planning.shortcuts import OneshotPlanner
from unified_planning.engines import PlanGenerationResultStatus

# Load domain and problem
reader = PDDLReader()
problem = reader.parse_problem("openrouter_domain.pddl", "openrouter_problem.pddl")

# Solve with a oneshot planner
with OneshotPlanner(problem_kind=problem.kind) as planner:
    result = planner.solve(problem)

# Print the plan if it was found
if result.status == PlanGenerationResultStatus.SOLVED_SATISFICING:
    print("✅ Plan found:")
    for action in result.plan.actions:
        print(action)
else:
    print("❌ No plan found.")
```

```python
from unified_planning.io import PDDLReader
from unified_planning.shortcuts import OneshotPlanner

reader = PDDLReader()
problem = reader.parse_problem("openrouter_domain.pddl", "openrouter_problem.pddl")

with OneshotPlanner(name="tamer",problem_kind=problem.kind) as planner:
    result = planner.solve(problem)
    if result.plan is not None:
        print("Plan:")
        for action in result.plan.actions:
            print(action)
    else:
        print("No plan found.")
```
